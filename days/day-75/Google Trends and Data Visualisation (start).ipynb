### Tesla Stock Price v.s. Search Volume

**Challenge:** Plot the Tesla stock price against the Tesla search volume using a line chart and two different axes. Label one axis 'TSLA Stock Price' and the other 'Search Trend'.

```python
df_tesla.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(14,10))

fontsize = 18

#Extract the axes
ax1 = plt.gca() # get current axes
ax2 = ax1.twinx() # create another axis sharing the same x1 axes

# Plots
ax1.plot(df_tesla['DATE'], df_tesla['TSLA_WEB_SEARCH'].values, color="green", linewidth=3)

ax2.plot(df_tesla['DATE'], df_tesla['TSLA_USD_CLOSE'].values, color="blue", linewidth=3)

# Set the minimum and maximum values on the axes
ax1.set_ylim([0, 600])
ax1.set_xlim([df_tesla.DATE.min(), df_tesla.DATE.max()])

# Labels
ax1.set_xlabel("Date", fontsize=fontsize)
ax1.set_ylabel("Web search", color="green", fontsize=fontsize)
ax2.set_ylabel("USD close", color="blue", fontsize=fontsize)
plt.title("Tesla", fontsize=fontsize)


# Improve date display
plt.xticks(rotation=45)

# Show
plt.tight_layout()
plt.show()
```
